```python
import os
import shutil
import random
import time
import copy
import io
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torchvision import transforms, datasets, models
from torch.utils.data import DataLoader
from PIL import Image
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

print("==========================================================")
print("🚀 RECYLINK AI - PERSON 1 COMPLETE PRODUCTION PIPELINE")
print("==========================================================\n")

# ---------------------------------------------------------
# STEP 1: AUTOMATED DATASET SPLIT (70% Train, 15% Val, 15% Test)
# ---------------------------------------------------------
SOURCE_DIR = "/content/epr_individual_images"
BASE_DIR = "/content/recylink_epr_dataset"
EPR_CLASSES = ["ITEW", "CEEW", "LSEEW", "EETW", "TLSEW", "MDW", "LIW", "NON_EWASTE"]

for split in ["train", "val", "test"]:
    for cls in EPR_CLASSES:
        os.makedirs(os.path.join(BASE_DIR, split, cls), exist_ok=True)

print("📦 Splitting EPR dataset into Train, Val, and Test...")

for cls in EPR_CLASSES:
    src_cls_dir = os.path.join(SOURCE_DIR, cls)
    if os.path.exists(src_cls_dir):
        images = [f for f in os.listdir(src_cls_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        random.shuffle(images)

        total = len(images)
        train_end = max(1, int(total * 0.70))
        val_end = max(train_end + 1, train_end + int(total * 0.15)) if total > 2 else train_end

        for img in images[:train_end]:
            shutil.copy(os.path.join(src_cls_dir, img), os.path.join(BASE_DIR, "train", cls, img))
        for img in images[train_end:val_end]:
            shutil.copy(os.path.join(src_cls_dir, img), os.path.join(BASE_DIR, "val", cls, img))
        for img in images[val_end:]:
            shutil.copy(os.path.join(src_cls_dir, img), os.path.join(BASE_DIR, "test", cls, img))

print("✅ Dataset splitting complete!")

# ---------------------------------------------------------
# STEP 2: DATALOADERS & PREPROCESSING PIPELINE
# ---------------------------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"⚡ Processing Engine Hardware: {device}\n")

BATCH_SIZE = 16

data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(degrees=15),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ]),
    'val': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ]),
    'test': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ]),
}

image_datasets = {
    x: datasets.ImageFolder(os.path.join(BASE_DIR, x), data_transforms[x])
    for x in ['train', 'val', 'test']
}

dataloaders = {
    x: DataLoader(image_datasets[x], batch_size=BATCH_SIZE, shuffle=(x == 'train'), num_workers=0)
    for x in ['train', 'val', 'test']
}

for split in ['train', 'val', 'test']:
    print(f"  • {split.upper()} Set Size: {len(image_datasets[split])} images")

# ---------------------------------------------------------
# STEP 3: MOBILENETV2 ARCHITECTURE INITIALIZATION
# ---------------------------------------------------------
weights = models.MobileNet_V2_Weights.DEFAULT
model = models.mobilenet_v2(weights=weights)

for param in model.parameters():
    param.requires_grad = False

in_features = model.classifier[1].in_features
model.classifier = nn.Sequential(
    nn.Dropout(p=0.2),
    nn.Linear(in_features, 128),
    nn.ReLU(),
    nn.Dropout(p=0.2),
    nn.Linear(128, len(EPR_CLASSES))
)

model = model.to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.classifier.parameters(), lr=0.001)

# ---------------------------------------------------------
# STEP 4: MODEL TRAINING LOOP (5 Epochs)
# ---------------------------------------------------------
num_epochs = 5
best_model_wts = copy.deepcopy(model.state_dict())
best_acc = 0.0

print("\n🚀 Training MobileNetV2 on 8 EPR Regulatory Classes...")

for epoch in range(num_epochs):
    print(f"\nEpoch {epoch+1}/{num_epochs}")
    print("-" * 25)

    for phase in ['train', 'val']:
        if phase == 'train':
            model.train()
        else:
            model.eval()

        running_loss = 0.0
        running_corrects = 0

        for inputs, labels in dataloaders[phase]:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()

            with torch.set_grad_enabled(phase == 'train'):
                outputs = model(inputs)
                _, preds = torch.max(outputs, 1)
                loss = criterion(outputs, labels)

                if phase == 'train':
                    loss.backward()
                    optimizer.step()

            running_loss += loss.item() * inputs.size(0)
            running_corrects += torch.sum(preds == labels.data)

        epoch_loss = running_loss / len(image_datasets[phase])
        epoch_acc = running_corrects.double() / len(image_datasets[phase])

        print(f"{phase.upper()} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}")

        if phase == 'val' and epoch_acc > best_acc:
            best_acc = epoch_acc
            best_model_wts = copy.deepcopy(model.state_dict())

print(f"\n🏆 Training Finished! Best Validation Accuracy: {best_acc:.4f}")

# Save trained weights
EXPORT_DIR = "/content/exported_model"
os.makedirs(EXPORT_DIR, exist_ok=True)
model.load_state_dict(best_model_wts)
model_save_path = os.path.join(EXPORT_DIR, "mobilenet_epr_classifier.pth")
torch.save(model.state_dict(), model_save_path)
print(f"✅ Trained weights saved at: `{model_save_path}`")

# ---------------------------------------------------------
# STEP 5: EVALUATION (Classification Report & Confusion Matrix)
# ---------------------------------------------------------
print("\n📊 Generating Classification Report & Confusion Matrix...")

model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for inputs, labels in dataloaders['test']:
        inputs = inputs.to(device)
        outputs = model(inputs)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.numpy())

class_names = [EPR_CLASSES[i] for i in sorted(image_datasets['test'].class_to_idx.values())]

print("\n--- CLASSIFICATION METRICS REPORT ---")
print(classification_report(all_labels, all_preds, target_names=class_names, zero_division=0))

cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted EPR Category')
plt.ylabel('Actual EPR Category')
plt.title('EPR E-Waste Classification - Confusion Matrix')
plt.show()

# ---------------------------------------------------------
# STEP 6: FASTAPI SERVER ENDPOINT CODE GENERATION (`app.py`)
# ---------------------------------------------------------
fastapi_code = """
from fastapi import FastAPI, File, UploadFile, HTTPException
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import transforms, models
from PIL import Image
import io
import os

app = FastAPI(title="RecyLink AI - EPR Classification API")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

EPR_CLASSES = ["ITEW", "CEEW", "LSEEW", "EETW", "TLSEW", "MDW", "LIW", "NON_EWASTE"]

EPR_DESCRIPTIONS = {
    "ITEW": "Information Technology and Telecommunication Equipment",
    "CEEW": "Consumer Electrical and Electronics and Photovoltaic Panels",
    "LSEEW": "Large Electrical and Electronic Equipment",
    "EETW": "Electrical and Electronic Tools",
    "TLSEW": "Toys, Leisure, and Sports Equipment",
    "MDW": "Medical Devices",
    "LIW": "Laboratory Instruments",
    "NON_EWASTE": "Non-E-Waste Materials (General Recyclables / Waste)"
}

def load_epr_model(weights_path="/content/exported_model/mobilenet_epr_classifier.pth"):
    weights = models.MobileNet_V2_Weights.DEFAULT
    model = models.mobilenet_v2(weights=weights)

    in_features = model.classifier[1].in_features
    model.classifier = nn.Sequential(
        nn.Dropout(p=0.2),
        nn.Linear(in_features, 128),
        nn.ReLU(),
        nn.Dropout(p=0.2),
        nn.Linear(128, len(EPR_CLASSES))
    )

    if os.path.exists(weights_path):
        model.load_state_dict(torch.load(weights_path, map_location=device))

    model.to(device)
    model.eval()
    return model

model = load_epr_model()

preprocess = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

@app.post("/predict")
async def predict_epr_category(file: UploadFile = File(...)):
    try:
        image_bytes = await file.read()
        raw_image = Image.open(io.BytesIO(image_bytes)).convert("RGB")
        input_tensor = preprocess(raw_image).unsqueeze(0).to(device)

        with torch.no_grad():
            outputs = model(input_tensor)
            probabilities = F.softmax(outputs, dim=1)
            conf, pred_idx = torch.max(probabilities, 1)

        predicted_code = EPR_CLASSES[pred_idx.item()]
        confidence_score = round(float(conf.item()), 4)

        return {
            "epr_code": predicted_code,
            "regulatory_category": EPR_DESCRIPTIONS.get(predicted_code, "Unknown"),
            "confidence": confidence_score
        }
    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Inference error: {str(e)}")
"""

with open("/content/app.py", "w") as f:
    f.write(fastapi_code)

print("\n🎉 Person 1 Deliverable Complete: `app.py` has been written to disk!")
```
